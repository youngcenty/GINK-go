```python
import re
import os
import pandas as pd
import matplotlib.pyplot as plt
from glob import glob
import numpy as np
from numpy import *
import seaborn as sns
import matplotlib.pyplot as plt
os.makedirs('result',exist_ok=True)
os.makedirs('trace',exist_ok=True)
os.makedirs('result/plot',exist_ok=True)
```

# HMM2+JHflag

```python
os.chdir("trace/")
files = glob('*withBg.trk')
for i in sorted(files):
    two = pd.read_csv(i,sep='\t')
    # figure1
    two.fillna(0)
    fig, ax1 = plt.subplots()
    color = 'tab:gray'
    ax1.set_xlabel('time')
    ax1.set_ylabel('intensity', color=color)
    ax1.plot(two['intensity'], color=color)
    ax1.tick_params(axis='y', labelcolor=color)
    ax2 = ax1.twinx()
    color = 'tab:blue'
    ax2.set_ylabel('2state', color=color)
    ax2.plot(two['hmm'], color=color)
    ax2.tick_params(axis='y', labelcolor=color)
    ax3 = ax1.twinx()
    plt.savefig(i.replace('.trk','_2color.png'))
    plt.show()
os.chdir("../")   
```
