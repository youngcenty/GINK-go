# HMM2+JHflag

```python
os.chdir("trace/")
files = glob('*_hmm2states.csv')
for i in sorted(files):
    if 1:
        two = pd.read_csv(i)
        if 'HMMStates' not in two.columns:continue
        # figure1
        two.fillna(0)
        fig, ax1 = plt.subplots()
        color = 'tab:gray'
        ax1.set_xlabel('time')
        ax1.set_ylabel('intensity', color=color)
        ax1.plot(two['photon_number'], color=color)
        ax1.tick_params(axis='y', labelcolor=color)
        ax2 = ax1.twinx()
        color = 'tab:blue'
        ax2.set_ylabel('2state', color=color)
        ax2.plot(two['HMMStates'], color=color)
        ax2.tick_params(axis='y', labelcolor=color)
        ax3 = ax1.twinx()
        color = 'tab:red'
        ax3.set_ylabel('2state', color=color)
        ax3.plot(two['TP_Flag'], color=color)
        ax3.tick_params(axis='y', labelcolor=color)
        plt.savefig(i.replace('.csv','_2color.png'))
        plt.show()
    
```
